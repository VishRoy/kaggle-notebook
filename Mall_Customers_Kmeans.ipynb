```python
df = pd.read_csv('/kaggle/input/datasets/vjchoudhary7/customer-segmentation-tutorial-in-python/Mall_Customers.csv')
df.head()
```

```python
print(df.info())
print(df.shape)
```

```python
df = df.drop('CustomerID',axis = 1)
df.head()
```

```python
df['Gender'] = df['Gender'].map({'Male': 0, 'Female': 1})
df.head()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df['Annual Income (k$)'], df['Spending Score (1-100)'], alpha=0.5)
plt.title("Raw Data - Annual Income vs Spending Score")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.show()
```

```python
plt.figure(figsize=(10, 6))  # bada plot
plt.scatter(df['Annual Income (k$)'], 
            df['Spending Score (1-100)'], 
            alpha=0.7, s=50)
plt.title("Raw Data - Annual Income vs Spending Score")
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.grid(True)  # grid lagao — groups aur clear dikhenge
plt.show()
```
